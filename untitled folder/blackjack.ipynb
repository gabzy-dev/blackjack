```python
import random

suits = ('Hearts', 'Diamonds', 'Spades', 'Clubs')
ranks = ('Two', 'Three', 'Four', 'Five', 'Six', 'Seven', 'Eight', 'Nine', 'Ten', 'Jack', 'Queen', 'King', 'Ace')
values = {'Two':2, 'Three':3, 'Four':4, 'Five':5, 'Six':6, 'Seven':7, 'Eight':8, 'Nine':9, 'Ten':10, 'Jack':10,
         'Queen':10, 'King':10, 'Ace':11}

playing = True 
```

```python
class Card: 
    def __init__(self,suit,rank): 
        self.suit = suit 
        self.rank = rank 

    def __str__(self): 
        return self.rank+ " of "+self.suit 

###Deck__cclass
class Deck: 
    def __init__(self): 
        self.deck = [] ##start the deck with an empty list 
        for suit in suits: 
            for rank in ranks: 
                self.deck.append(Card(suit,rank))
    def __str__(self): 
        deck_comp = ' '
        for card in self.deck: 
            deck_comp += '\n' + card.__str__() 
        return f"the deck has {deck_comp}"

#####shuffling_the_deck 
    
    def shuffle(self): 
        random.shuffle(self.deck) 

    def deal(self): 
        single_card = self.deck.pop() 
        return single_card 

        
        
```

```python
class Hand:
    def __init__(self): 
        self.cards = [] ##you start the list of cards with an empty list
        self.value = 0  ##set the values of your card to zero for a start
        self.aces = 0    ##set your aces to zero as well 

    def add_card(self,card): 
        self.cards.append(card)
        self.value += values[card.rank]

        if card.rank == 'Ace': 
            self.aces += 1 
            

    def adjust_for_ace(self): 
        while self.value > 21 and self.aces > 0: 
            self.value -= 10
            self.aces  -= 1 
        

```

```python
###creating the chips class 
class chips: 
    def __init__(self,total = 100): 
        self.total = total
        self.bet = 0 

    def win_bet(self):
        self.total += self.bet 

    def loose_bet(self):  
        self.total -= self.bet 
```

```python
########function/ logic that plays the game 

def take_bet(chips): 
    while True: 
        try:
            chips.bet = int(input('how many chips will you like to play today ?'))
        except: 
            print('sorry you did not imput an integer')
        else:
            if chips.bet > chips.total: 
                print(f"sorry your bet is higher than your total value, you have{chips.total}left")
            else: 
                break 
            
             
```

```python
###function to take hits 
def hit(deck,hand): 
    single_card = deck.deal() 
    hand.add_card(single_card) 
    hand.adjust_for_ace()
```

```python
######function asking if a player will take hits or stand
def hit_or_stand(deck,hand): 
    global playing
    
    while True: 
        x = input('hit or stand ? enter h or s')
        if x[0].lower() == 'h': 
            hit(deck,hand)

        elif x[0].lower() == 's':  
            print('player decides to stand, its dealers turn.')
            playing = False 

        else: 
            print('please select either h to hit or s to stay')
            continue ##continue here makes the hit or stand function starts again after the player selects h to hit  or s to stay
        break 
        
        
```
